#Country Translation Lookup Notebook

```python
#Import packages
import pandas as pd
```

```python
#Load raw data
country_df = pd.read_csv('country-table-export.csv')
cldr_df = pd.read_csv('cldr_names.csv')
```

```python
# Map BCP47 codes to CLDR columns
bcp47_to_cldr = {
    'fr-fr': 'cldr_name_fr_FR',
    'de-de': 'cldr_name_de_DE',
    'es-es': 'cldr_name_es_ES',
    'it-it': 'cldr_name_it_IT',
    'fi': 'cldr_name_fi',
    'nn-no': 'cldr_name_nn_NO',
    'nb-no': 'cldr_name_nb_NO',
    'sv-se': 'cldr_name_sv_SE',
    'da-dk': 'cldr_name_da_DK',
    'zh': 'cldr_name_zh',
    'pl-pl': 'cldr_name_pl_PL',
    'ko': 'cldr_name_ko',
    'cs-cz': 'cldr_name_cs_CZ',
    'ja-jp': 'cldr_name_ja_JP',
    'nl-nl': 'cldr_name_nl_NL',
    'pt-pt': 'cldr_name_pt_PT'
}
```

```python
# Merge on ISO31661A2CountryCode <-> territory_code
merged = country_df.merge(
    cldr_df,
    left_on='ISO31661A2CountryCode',
    right_on='territory_code',
    how='left'
)
```

```python
# Select columns for output
output_cols = [
    'ISO31661A2CountryCode', 'CountryEnglishName'
] + [bcp47_to_cldr[bcp] for bcp in bcp47_to_cldr]

output_df = merged[output_cols]
output_df.columns = (
    ['ISO31661A2CountryCode', 'CountryEnglishName'] +
    list(bcp47_to_cldr.keys())
)
```

```python
# Save to CSV for further use
output_df.to_csv('country_translations_lookup_output.csv', index=False, encoding='utf-8')
```
